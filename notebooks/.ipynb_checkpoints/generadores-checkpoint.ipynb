# Generadores de Variables Aleatorias — Simulación de Llegadas y Compras de Clientes

Este notebook valida, de forma empírica, los tres generadores implementados en el motor de Rust
(`random_engine`, expuesto a Python vía `pyo3`):

1. **Proceso de Poisson** para las llegadas de clientes (tiempos entre llegadas vía **Transformada Inversa** sobre la exponencial).
2. **Método de Composición** para la cantidad comprada por cliente (mezcla 90% regular / 10% mayorista).
3. **Comparación rigurosa** entre Transformada Inversa y Aceptación-Rechazo para la cantidad del cliente regular.

Para cada generador se valida media, varianza y bondad de ajuste (Chi-cuadrado) contra la distribución
teórica declarada, y se cierra con una comparación cuantitativa de eficiencia entre Inversa y Rechazo.

```python
import time

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy import stats

import random_engine as re

np.random.seed(42)
plt.rcParams["figure.figsize"] = (8, 4.5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3

resumen = []  # aquí se acumula la tabla de validación estadística de la sección 4
```

## 1. Proceso de Poisson — llegadas de clientes

Las llegadas de clientes a la tienda forman un Proceso de Poisson de tasa $\lambda$ (clientes por minuto).
Bajo este modelo, los **tiempos entre llegadas** $T$ son variables Exponenciales$(\lambda)$, con densidad

$$f(t) = \lambda e^{-\lambda t}, \qquad t \ge 0$$

y se generan con el **Método de la Transformada Inversa**:

$$T = -\frac{\ln(1-U)}{\lambda}, \qquad U \sim \text{Uniforme}(0,1)$$

que es exactamente lo que implementa `poisson_arrival(rate)` en Rust (`engines::poisson_process_arrival` →
`util::inverse_exponential`).

```python
lambda_llegadas = 3.0  # clientes por minuto -> tiempo medio entre llegadas = 1/3 min = 20 s
N1 = 20_000

t_entre_llegadas = np.array([re.poisson_arrival(lambda_llegadas) for _ in range(N1)])

media_obs = t_entre_llegadas.mean()
var_obs = t_entre_llegadas.var(ddof=1)
media_teo = 1 / lambda_llegadas
var_teo = 1 / lambda_llegadas**2

print(f"Media muestral   : {media_obs:.5f}  |  Media teórica   : {media_teo:.5f}")
print(f"Varianza muestral: {var_obs:.5f}  |  Varianza teórica: {var_teo:.5f}")
```

```python
# Bondad de ajuste (Chi-cuadrado) con bins de igual probabilidad bajo Exponencial(lambda)
k_bins = 15
bordes = stats.expon(scale=1 / lambda_llegadas).ppf(np.linspace(0, 1, k_bins + 1))
bordes[-1] = np.inf

obs, _ = np.histogram(t_entre_llegadas, bins=bordes)
esp = np.full(k_bins, N1 / k_bins)

chi2_stat, p_valor = stats.chisquare(obs, esp)
print(f"Chi-cuadrado = {chi2_stat:.3f}  (gl={k_bins - 1}),  p-valor = {p_valor:.4f}")
print("No se rechaza H0 (Exponencial)" if p_valor > 0.05 else "Se rechaza H0 (Exponencial)")

resumen.append({
    "generador": "Llegadas (Inversa)",
    "distribución": f"Exp(λ={lambda_llegadas})",
    "N": N1,
    "media_obs": media_obs, "media_teo": media_teo,
    "var_obs": var_obs, "var_teo": var_teo,
    "chi2": chi2_stat, "p_valor": p_valor,
})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

x = np.linspace(0, np.percentile(t_entre_llegadas, 99.5), 300)
axes[0].hist(t_entre_llegadas, bins=60, density=True, alpha=0.6, color="#4C72B0", label="Muestra")
axes[0].plot(x, stats.expon(scale=1 / lambda_llegadas).pdf(x), color="#C44E52", lw=2, label="Exp teórica")
axes[0].set_title("Tiempos entre llegadas (Transformada Inversa)")
axes[0].set_xlabel("t (minutos)")
axes[0].set_ylabel("densidad")
axes[0].legend()

# Validación cruzada: acumular tiempos entre llegadas -> conteo de llegadas en una ventana fija
ventana = 30.0  # minutos
llegadas_acum = np.cumsum(t_entre_llegadas)
n_simulaciones = 3000
conteos = []
for _ in range(n_simulaciones):
    ts = np.cumsum(np.array([re.poisson_arrival(lambda_llegadas) for _ in range(200)]))
    conteos.append(np.sum(ts <= ventana))
conteos = np.array(conteos)

k_max = conteos.max()
ks = np.arange(0, k_max + 1)
axes[1].hist(conteos, bins=np.arange(-0.5, k_max + 1.5, 1), density=True, alpha=0.6,
             color="#4C72B0", label="Conteo simulado")
axes[1].plot(ks, stats.poisson(mu=lambda_llegadas * ventana).pmf(ks), "o-", color="#C44E52",
             label="Poisson teórica")
axes[1].set_title(f"# de llegadas en ventana de {ventana:.0f} min")
axes[1].set_xlabel("# llegadas")
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"Media conteo llegadas/ventana: obs={conteos.mean():.3f}  teo={lambda_llegadas*ventana:.3f}")
```

**Lectura:** el generador de tiempos entre llegadas por Transformada Inversa reproduce la Exponencial
declarada (media/varianza cercanas a las teóricas, prueba Chi-cuadrado no significativa en esta corrida),
y al acumular esos tiempos, el conteo de llegadas en una ventana fija reproduce la Poisson teórica esperada
— confirmando que el generador de tiempos entre llegadas efectivamente construye un Proceso de Poisson
válido. (Si en alguna corrida el Chi-cuadrado rechazara H0, ver la nota sobre error Tipo I en la sección 4.)

## 2. Método de Composición — cantidad comprada por cliente

La cantidad comprada por un cliente es una mezcla de dos poblaciones:

- **90%** clientes regulares, cantidad $\sim \text{Exponencial}(\lambda_r)$
- **10%** clientes mayoristas, cantidad $\sim \text{Exponencial}(\lambda_m)$, con $\lambda_m < \lambda_r$ (compran más)

El **Método de Composición** primero decide con $U_1$ a qué subpoblación pertenece el cliente
(90%/10%) y luego genera la cantidad con Transformada Inversa sobre la exponencial correspondiente.
Esto es lo que implementa `customer_quantity(regular_rate, wholesale_rate)`.

```python
lambda_reg = 1.0   # cantidad media regular = 1 kg
lambda_may = 0.1   # cantidad media mayorista = 10 kg
N2 = 20_000

muestras = [re.customer_quantity(lambda_reg, lambda_may) for _ in range(N2)]
cantidades = np.array([q for q, _ in muestras])
es_mayorista = np.array([w for _, w in muestras])

p_mayorista_obs = es_mayorista.mean()
print(f"Proporción mayorista observada: {p_mayorista_obs:.4f}  (teórica: 0.10)")

test_binom = stats.binomtest(int(es_mayorista.sum()), N2, p=0.10)
print(f"Prueba binomial de la proporción -> p-valor = {test_binom.pvalue:.4f}")
```

```python
cant_reg = cantidades[~es_mayorista]
cant_may = cantidades[es_mayorista]

def validar_subpoblacion(nombre, muestra, lam):
    media_obs, var_obs = muestra.mean(), muestra.var(ddof=1)
    media_teo, var_teo = 1 / lam, 1 / lam**2

    k_bins = 10
    bordes = stats.expon(scale=1 / lam).ppf(np.linspace(0, 1, k_bins + 1))
    bordes[-1] = np.inf
    obs, _ = np.histogram(muestra, bins=bordes)
    esp = np.full(k_bins, len(muestra) / k_bins)
    chi2_stat, p_valor = stats.chisquare(obs, esp)

    print(f"[{nombre}] media obs={media_obs:.4f} teo={media_teo:.4f} | "
          f"var obs={var_obs:.4f} teo={var_teo:.4f} | chi2={chi2_stat:.3f} p={p_valor:.4f}")

    resumen.append({
        "generador": f"Composición - {nombre}",
        "distribución": f"Exp(λ={lam})",
        "N": len(muestra),
        "media_obs": media_obs, "media_teo": media_teo,
        "var_obs": var_obs, "var_teo": var_teo,
        "chi2": chi2_stat, "p_valor": p_valor,
    })

validar_subpoblacion("regular", cant_reg, lambda_reg)
validar_subpoblacion("mayorista", cant_may, lambda_may)

media_mezcla_teo = 0.9 / lambda_reg + 0.1 / lambda_may
media_mezcla_obs = cantidades.mean()
print(f"\nMedia de la mezcla completa: obs={media_mezcla_obs:.4f}  teo={media_mezcla_teo:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hist(cant_reg, bins=60, density=False, weights=np.full(len(cant_reg), 1 / N2),
        alpha=0.6, color="#4C72B0", label="Regular (90%)")
ax.hist(cant_may, bins=60, density=False, weights=np.full(len(cant_may), 1 / N2),
        alpha=0.6, color="#DD8452", label="Mayorista (10%)")
ax.set_title("Cantidad comprada por cliente — Método de Composición")
ax.set_xlabel("cantidad (kg)")
ax.set_ylabel("proporción de la muestra")
ax.set_xlim(0, np.percentile(cantidades, 99))
ax.legend()
plt.tight_layout()
plt.show()
```

**Lectura:** la proporción observada de clientes mayoristas ronda el 10% declarado (prueba binomial
no significativa), y cada subpoblación reproduce de cerca la Exponencial declarada, con media y varianza
muestral próximas a los valores teóricos — el Método de Composición está generando correctamente la mezcla.
Una prueba Chi-cuadrado puntual puede rechazar por azar (ver nota de error Tipo I en la sección 4).

## 3. Comparación rigurosa: Transformada Inversa vs. Aceptación-Rechazo

Ahora se genera la **misma variable** — la cantidad del cliente regular, Exponencial($\lambda$) — con dos
métodos distintos, para compararlos:

- **Transformada Inversa**: $X = -\ln(1-U)/\lambda$ (`regular_quantity_inverse`).
- **Aceptación-Rechazo** con envolvente Exponencial de tasa $\mu = \lambda/2$ (`regular_quantity_accept_reject`).

Derivación de la constante envolvente: con $f(x)=\lambda e^{-\lambda x}$ y $g(x)=\mu e^{-\mu x}$,

$$\frac{f(x)}{g(x)} = \frac{\lambda}{\mu} e^{-(\lambda-\mu)x}$$

que se maximiza en $x=0$, dando $c = \lambda/\mu = 2$. La tasa de aceptación teórica es $1/c = 50\%$,
y el criterio de aceptación se reduce a $U \le e^{-\mu Y}$.

```python
lambda_3 = lambda_reg  # misma tasa que el cliente regular de la sección 2
N3 = 20_000

t0 = time.perf_counter()
muestras_inv = np.array([re.regular_quantity_inverse(lambda_3) for _ in range(N3)])
t_inv_N3 = time.perf_counter() - t0

t0 = time.perf_counter()
resultados_ar = [re.regular_quantity_accept_reject(lambda_3) for _ in range(N3)]
t_ar_N3 = time.perf_counter() - t0

muestras_ar = np.array([v for v, _ in resultados_ar])
intentos_ar = np.array([a for _, a in resultados_ar])

tasa_aceptacion_obs = N3 / intentos_ar.sum()
print(f"Tasa de aceptación observada: {tasa_aceptacion_obs:.4f}  (teórica: 0.50)")
print(f"Intentos promedio por muestra aceptada: {intentos_ar.mean():.4f}  (teórico: 2.00)")
```

```python
def validar_metodo(nombre, muestra, lam):
    media_obs, var_obs = muestra.mean(), muestra.var(ddof=1)
    media_teo, var_teo = 1 / lam, 1 / lam**2

    k_bins = 15
    bordes = stats.expon(scale=1 / lam).ppf(np.linspace(0, 1, k_bins + 1))
    bordes[-1] = np.inf
    obs, _ = np.histogram(muestra, bins=bordes)
    esp = np.full(k_bins, len(muestra) / k_bins)
    chi2_stat, p_valor = stats.chisquare(obs, esp)

    print(f"[{nombre}] media obs={media_obs:.4f} teo={media_teo:.4f} | "
          f"var obs={var_obs:.4f} teo={var_teo:.4f} | chi2={chi2_stat:.3f} p={p_valor:.4f}")

    resumen.append({
        "generador": nombre,
        "distribución": f"Exp(λ={lam})",
        "N": len(muestra),
        "media_obs": media_obs, "media_teo": media_teo,
        "var_obs": var_obs, "var_teo": var_teo,
        "chi2": chi2_stat, "p_valor": p_valor,
    })

validar_metodo("Cantidad regular (Inversa)", muestras_inv, lambda_3)
validar_metodo("Cantidad regular (Rechazo)", muestras_ar, lambda_3)

test_acept = stats.binomtest(N3, int(intentos_ar.sum()), p=0.5)
print(f"\nPrueba binomial tasa de aceptación (H0: p=0.5) -> p-valor = {test_acept.pvalue:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
x = np.linspace(0, np.percentile(muestras_inv, 99.5), 300)
ax.hist(muestras_inv, bins=60, density=True, alpha=0.5, color="#4C72B0", label="Inversa")
ax.hist(muestras_ar, bins=60, density=True, alpha=0.5, color="#55A868", label="Rechazo")
ax.plot(x, stats.expon(scale=1 / lambda_3).pdf(x), color="#C44E52", lw=2, label="Exp teórica")
ax.set_title("Cantidad del cliente regular — Inversa vs. Rechazo")
ax.set_xlabel("cantidad (kg)")
ax.set_ylabel("densidad")
ax.legend()
plt.tight_layout()
plt.show()
```

**Lectura:** ambos métodos generan muestras estadísticamente indistinguibles de la Exponencial
declarada (medias, varianzas y Chi-cuadrado consistentes), y la tasa de aceptación observada del método
de Rechazo ronda el 50% teórico esperado — validando la derivación de la constante envolvente $c=2$.

## 4. Validación estadística consolidada

Tabla resumen de media, varianza y prueba Chi-cuadrado de bondad de ajuste para todos los generadores
evaluados en las secciones anteriores.

> **Nota sobre error Tipo I:** este notebook corre ~7 pruebas Chi-cuadrado/binomiales independientes con
> $\alpha=0.05$. Por definición, si todos los generadores son correctos, se espera igual que ≈5% de esas
> pruebas rechacen $H_0$ por puro azar (falso positivo). Por eso el criterio de validación no es "¿pasó
> el Chi-cuadrado?" de forma aislada, sino el conjunto: ¿la media y varianza muestral están cerca de las
> teóricas, y la gran mayoría de las pruebas no rechazan? Un rechazo aislado con media/varianza casi
> idénticas a la teórica es exactamente el comportamiento esperado de la prueba, no evidencia de un
> generador defectuoso.

```python
df_resumen = pd.DataFrame(resumen)
df_resumen["error_media_%"] = 100 * (df_resumen.media_obs - df_resumen.media_teo).abs() / df_resumen.media_teo
df_resumen["error_var_%"] = 100 * (df_resumen.var_obs - df_resumen.var_teo).abs() / df_resumen.var_teo
df_resumen["pasa_chi2 (α=0.05)"] = df_resumen.p_valor > 0.05
df_resumen.round(4)
```

## 5. Comparación cuantitativa: Inversa vs. Rechazo

Se mide el tiempo de generación de $N$ muestras con cada método y se cuantifica el número de candidatos
generados/descartados por el método de Rechazo.

```python
N5 = 50_000

t0 = time.perf_counter()
_ = [re.regular_quantity_inverse(lambda_3) for _ in range(N5)]
t_inversa = time.perf_counter() - t0

t0 = time.perf_counter()
resultados_5 = [re.regular_quantity_accept_reject(lambda_3) for _ in range(N5)]
t_rechazo = time.perf_counter() - t0

intentos_5 = np.array([a for _, a in resultados_5])
candidatos_generados = intentos_5.sum()
candidatos_descartados = candidatos_generados - N5

print(f"N = {N5:,} muestras aceptadas en ambos métodos\n")
print(f"Tiempo Inversa : {t_inversa:.4f} s  ->  {N5 / t_inversa:,.0f} muestras/s")
print(f"Tiempo Rechazo : {t_rechazo:.4f} s  ->  {N5 / t_rechazo:,.0f} muestras/s")
print(f"Rechazo / Inversa (tiempo): {t_rechazo / t_inversa:.2f}x más lento\n")
print(f"Candidatos generados (Rechazo)  : {candidatos_generados:,}")
print(f"Candidatos descartados (Rechazo): {candidatos_descartados:,}  "
      f"({100 * candidatos_descartados / candidatos_generados:.2f}% del total generado)")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].bar(["Inversa", "Rechazo"], [t_inversa, t_rechazo], color=["#4C72B0", "#55A868"])
axes[0].set_ylabel("segundos")
axes[0].set_title(f"Tiempo de generación de {N5:,} muestras")
for i, v in enumerate([t_inversa, t_rechazo]):
    axes[0].text(i, v, f"{v:.3f}s", ha="center", va="bottom")

axes[1].bar(["Aceptados", "Descartados"], [N5, candidatos_descartados],
            color=["#55A868", "#C44E52"])
axes[1].set_title("Candidatos generados por el método de Rechazo")
axes[1].set_ylabel("# candidatos")
for i, v in enumerate([N5, candidatos_descartados]):
    axes[1].text(i, v, f"{v:,}", ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

## 6. Conclusión

Con los datos generados en este notebook:

- **Corrección:** ambos métodos (Inversa y Rechazo) producen muestras estadísticamente compatibles con la
  Exponencial declarada — mismas medias, varianzas y pruebas Chi-cuadrado no significativas (sección 4).
- **Tasa de aceptación:** el método de Rechazo acepta aproximadamente el 50% de los candidatos generados,
  tal como predice la constante envolvente $c=\lambda/\mu=2$ derivada en la sección 3.
- **Eficiencia:** el método de Rechazo descarta cerca de la mitad de los candidatos que genera (sección 5),
  y por cada muestra requiere en promedio **dos** evaluaciones de exponencial más un uniforme adicional
  para el criterio de aceptación, contra **una sola** evaluación en la Transformada Inversa. Esto se
  traduce directamente en un mayor tiempo de cómputo para el mismo número de muestras aceptadas.

**Recomendación:** para el modelo final se recomienda la **Transformada Inversa**. Cuando la exponencial
tiene una inversa cerrada disponible — como es el caso aquí — la Transformada Inversa siempre acepta,
usa un solo uniforme por muestra y es más rápida que el Rechazo, que además introduce varianza extra en el
costo computacional (número de intentos por muestra es geométrico). El método de Aceptación-Rechazo solo
se justificaría si la distribución objetivo no tuviera una inversa fácil de calcular, que no es el caso de
la Exponencial usada para modelar la cantidad comprada por el cliente regular.
